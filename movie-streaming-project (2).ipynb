# Q1:
# List all Movies released in 2000 or later, newest first , with title / release year / duration :

```python
query = """
SELECT
    Title,
    Year,
    Runtime
FROM movies
WHERE Year >= 2000
ORDER BY Year DESC;
"""

q1 = pd.read_sql_query(query, conn)
q1
```

# Q2:
# How many subscribers are on each plan + what is the average watch time for each plan?

```python
query = """
SELECT
    Subscription_Type,
    COUNT(*) AS subscriber_count,
    ROUND(AVG(Watch_Time_Hours), 2) AS average_watch_hours
FROM users
GROUP BY Subscription_Type
ORDER BY subscriber_count DESC;
"""

q2 = pd.read_sql_query(query, conn)
q2
```

# Q3:
# Find the 5 longest and 5 shortest movies :

```python
# Longest
query = """
SELECT
    Title,
    Runtime
FROM movies
WHERE Runtime IS NOT NULL
ORDER BY Runtime DESC
LIMIT 5;
"""

q3_longest = pd.read_sql_query(query, conn)
q3_longest
```

```python
# Shortest

query = """
SELECT
    Title,
    Runtime
FROM movies
WHERE Runtime IS NOT NULL
ORDER BY Runtime ASC
LIMIT 5;
"""

q3_shortest = pd.read_sql_query(query, conn)
q3_shortest
```

# Q4:
# Which distinct countries do our subscribers come from + how many subscribers are from each country?

```python
query = """
SELECT
    Country,
    COUNT(*) AS subscriber_count
FROM users
GROUP BY Country
ORDER BY subscriber_count DESC;
"""

q4 = pd.read_sql_query(query, conn)
q4
```

# Q5:
#  What is the Average watch time by favorite genre?

```python
query = """
SELECT
    Favorite_Genre,
    COUNT(*) AS subscriber_count,
    ROUND(AVG(Watch_Time_Hours), 2) AS average_watch_hours
FROM users
GROUP BY Favorite_Genre
ORDER BY average_watch_hours DESC;
"""

q5 = pd.read_sql_query(query, conn)
q5
```

---


# Q6:
# What share of users belong to each subscription type, and what is the average watch time per subscription type?

```python
query6 = """
SELECT
    Subscription_Type,
    COUNT(*) AS user_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM users),
        2
    ) AS user_share_percentage,
    ROUND(AVG(Watch_Time_Hours), 2) AS average_watch_time
FROM users
GROUP BY Subscription_Type;
"""

result6 = pd.read_sql_query(query6, conn)

result6
```

# Q7:
#  What is the distribution of IMDb ratings? What is the overall average rating?

```python
query7 = """
SELECT
    rating,
    COUNT(*) AS rating_count
FROM movies
WHERE rating IS NOT NULL
GROUP BY rating
ORDER BY rating;
"""

result7 = pd.read_sql_query(query7, conn)

query7_avg = """
SELECT
    ROUND(AVG(rating), 2) AS overall_average_rating
FROM movies
WHERE rating IS NOT NULL;
"""

result7_avg = pd.read_sql_query(query7_avg, conn)

print("IMDb Rating Distribution:")
display(result7)

print("\nOverall Average IMDb Rating:")
display(result7_avg)
```

# Q8:
# Which release years have at least 10 movies in the catalogue, and what is the average duration of their movies? Use HAVING.

```python
query8 = """
SELECT
    year,
    COUNT(*) AS movie_count,
    ROUND(AVG(runtime), 2) AS average_duration
FROM movies
WHERE kind = 'movie'
    AND year IS NOT NULL
    AND runtime IS NOT NULL
GROUP BY year
HAVING COUNT(*) >= 10
ORDER BY movie_count DESC;
"""

result8 = pd.read_sql_query(query8, conn)

result8
```

# Q9:
# Which 10 movies have the highest number of IMDb votes? Show title, genre and release year.

```python
query9 = """
SELECT
    title,
    genre,
    year AS release_year,
    vote AS number_of_votes
FROM movies
WHERE vote IS NOT NULL
ORDER BY vote DESC
LIMIT 10;
"""

result9 = pd.read_sql_query(query9, conn)

result9
```

# Q10:
# Rank the genres by number of movies and by average IMDb rating.

```python
query10 = """
SELECT
    genre,
    COUNT(*) AS movie_count,
    ROUND(AVG(rating), 2) AS average_rating
FROM movies
WHERE genre IS NOT NULL
  AND rating IS NOT NULL
GROUP BY genre
ORDER BY movie_count DESC;
"""

result10 = pd.read_sql_query(query10, conn)

result10
```


# Q11:
# Which are the 10 best-rated movies that have at least 10 IMDb votes? Show title, genre, number of votes and rating.

```python
query11 = """
SELECT
    title,
    genre,
    vote AS number_of_votes,
    rating
FROM movies
WHERE rating IS NOT NULL
  AND vote >= 10
ORDER BY rating DESC, vote DESC
LIMIT 10;
"""

result11 = pd.read_sql_query(query11, conn)

result11
```

# Q12:
#  What is the average rating per movie type? Which movie type has the lowest average rating

```python
query12 = """
SELECT
    kind,
    ROUND(AVG(rating), 2) AS average_rating
FROM movies
WHERE kind IS NOT NULL
  AND rating IS NOT NULL
GROUP BY kind
ORDER BY average_rating ASC;
"""

result12 = pd.read_sql_query(query12, conn)

result12
```

# Q13:
#  Who are the 10 heaviest viewers? Show name, subscription type, country and total hours watched.

```python
query13 = """
SELECT
    Name,
    Subscription_Type,
    Country,
    Watch_Time_Hours AS total_hours_watched
FROM users
ORDER BY Watch_Time_Hours DESC
LIMIT 10;
"""

result13 = pd.read_sql_query(query13, conn)

result13
```

# Q14:
#  Compare the plans: average hours watched per subscriber for Basic vs Standard vs Premium.

```python
query14 = """
SELECT
    Subscription_Type,
    COUNT(*) AS subscriber_count,
    ROUND(AVG(Watch_Time_Hours), 2) AS average_hours_watched
FROM users
GROUP BY Subscription_Type
ORDER BY average_hours_watched DESC;
"""

result14 = pd.read_sql_query(query14, conn)

result14
```

# Q15:

# Q16:

# Q17:
# What is the average IMDb rating by movie type? Which movie types have the highest average rating?

```python
query17 = """
SELECT 
    kind,
    COUNT(*) AS title_count,
    ROUND(AVG(rating), 2) AS average_rating
FROM movies
WHERE kind IS NOT NULL 
  AND rating IS NOT NULL
GROUP BY kind
ORDER BY average_rating DESC;
"""

result17 = pd.read_sql_query(query17, conn)
result17
```

# Q18:
# Bucket the catalogue into 'Classic' (before 2005), 'Modern' (2005–2017) and 'Recent' (2018+) and compare the number of titles and average rating for each bucket.

```python

query18 = """
SELECT
    CASE
        WHEN year < 2005 THEN 'Classic'
        WHEN year BETWEEN 2005 AND 2017 THEN 'Modern'
        WHEN year >= 2018 THEN 'Recent'
    END AS movie_bucket,
    COUNT(*) AS title_count,
    ROUND(AVG(rating), 2) AS average_rating
FROM movies
WHERE year IS NOT NULL
GROUP BY movie_bucket
ORDER BY movie_bucket;
"""

result18 = pd.read_sql_query(query18, conn)

result18
```

# Q19:
# Using a CTE and a window function, find the highest-rated movie in each movie type.

```python

query19 = """
WITH ranked_movies AS (
    SELECT 
        title,
        kind,
        rating,
        vote,
        ROW_NUMBER() OVER (
            PARTITION BY kind 
            ORDER BY rating DESC, vote DESC
        ) AS rn
    FROM movies
    WHERE kind IS NOT NULL 
      AND rating IS NOT NULL
)
SELECT 
    title,
    kind,
    rating,
    vote
FROM ranked_movies
WHERE rn = 1
ORDER BY kind;
"""

result19 = pd.read_sql_query(query19, conn)
result19
```

# Q20:
# Which subscribers have a watch time higher than the overall average watch time? Show their name, subscription type and watch time. Use a subquery.

```python

query20 = """
SELECT
    Name,
    Subscription_Type,
    Watch_Time_Hours
FROM users
WHERE Watch_Time_Hours > (
    SELECT AVG(Watch_Time_Hours)
    FROM users
)
ORDER BY Watch_Time_Hours DESC;
"""

result20 = pd.read_sql_query(query20, conn)

result20
```
